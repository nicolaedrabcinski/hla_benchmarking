```python
from collections import Counter
from pathlib import Path
import sys

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

project_folder = Path(
    r"C:\Users\annag\Desktop\PHD\hla_benchmarking"
)

datasets_folder = project_folder / "datasets"
results_folder = project_folder / "results" / "standard"


allele_colors = {
    'A': '#e41a1c',
    'B': '#377eb8',
    'C': '#4daf4a',
    'DQB1': '#984ea3',
    'DRB1': '#ff7f00',
}

tool_predicted_alleles = {

    "arcas": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "optitype": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1"
    ],

    "seq2hla": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "hlaminer": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "phlat": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "hlavbseq": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "hlahd": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "hisat": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "hlaforest": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "hlapers": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "T1K": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],

    "rna2hla": [
        "A", "A.1",
        "B", "B.1",
        "C", "C.1",
        "DQB1", "DQB1.1",
        "DRB1", "DRB1.1"
    ],
}

def normalize_allele_resolution(gs_allele, predicted_allele):
    """
    Normalize two alleles to the lower resolution of the two.

    Example:
        GS:         A*11:01:01
        Prediction: A*11:01

        Both are compared as:
        A*11:01

    The locus is always preserved.
    """

    gs_allele = str(gs_allele).strip()
    predicted_allele = str(predicted_allele).strip()

    if "*" not in gs_allele or "*" not in predicted_allele:
        return gs_allele, predicted_allele

    gs_locus, gs_fields = gs_allele.split("*", 1)
    pred_locus, pred_fields = predicted_allele.split("*", 1)

    # Different loci can never match
    if gs_locus != pred_locus:
        return gs_allele, predicted_allele

    gs_fields = gs_fields.split(":")
    pred_fields = pred_fields.split(":")

    # Use the lower resolution
    resolution = min(
        len(gs_fields),
        len(pred_fields)
    )

    gs_normalized = (
        gs_locus
        + "*"
        + ":".join(gs_fields[:resolution])
    )

    pred_normalized = (
        pred_locus
        + "*"
        + ":".join(pred_fields[:resolution])
    )

    return gs_normalized, pred_normalized

def get_locus(allele):

    if allele.startswith("DRB1*"):
        return "DRB1"

    if allele.startswith("DQB1*"):
        return "DQB1"

    if "*" in allele:
        return allele.split("*")[0]

    return None

# compare predictions with the gold standard
def get_miscalled_and_all_alleles(
    predictions,
    gs,
    tool_name,
    tool_predicted_alleles,
):
    """
    Compare predictions with the gold standard.

    - GS cells can contain multiple acceptable alleles separated by "/".
    - Prediction is correct if ANY predicted allele matches ANY allele in the GS cell.
    - NaN / empty prediction is a miscall ONLY for columns that the particular tool is supposed to predict.
    - Columns not predicted by the tool are ignored.
    """

    loci = [
        "A",
        "B",
        "C",
        "DRB1",
        "DQB1",
    ]

    miscalled_alleles = Counter()
    allele_opportunities = Counter()

    total_calls = 0
    miscalled_calls = 0

    predicted_columns = set(
        tool_predicted_alleles[tool_name]
    )

    if "Run" not in gs.columns:
        raise ValueError(
            "Gold standard must have 'Run' column."
        )

    prediction_id_col = next(
        (column for column in ("ERR", "Run") if column in predictions.columns),
        None,
    )
    if prediction_id_col is None:
        raise ValueError(
            "Predictions must have an 'ERR' or 'Run' column."
        )

    gs_index = gs.set_index("Run") # index by 'Run' for gold standard
    pred_index = predictions.set_index(prediction_id_col)

    common_runs = gs_index.index.intersection(
        pred_index.index
    )


    for run in common_runs:

        gs_row = gs_index.loc[run]
        pred_row = pred_index.loc[run]

        for locus in loci:

            columns = [
                locus,
                f"{locus}.1",
            ]

            for col in columns:
                if col not in predicted_columns:
                    continue
                if col not in gs.columns:
                    continue
                if col not in predictions.columns:
                    continue

                gs_value = gs_row[col]

                if pd.isna(gs_value):
                    continue

                gs_value = str(gs_value).strip()

                if gs_value in {
                    "",
                    ".",
                    "nan",
                    "NaN",
                    "None",
                }:
                    continue

                # because GS can contain multiple alleles, we split them and count each unique allele as an opportunity for miscall.
                gs_alleles = {
                    allele.strip()
                    for allele in gs_value.split("/")
                    if allele.strip()
                    and allele.strip() not in {
                        ".",
                        "nan",
                        "NaN",
                        "None",
                    }
                }

                if not gs_alleles:
                    continue

                total_calls += 1

                # Every unique GS allele gets one opportunity
                for allele in gs_alleles:
                    allele_opportunities[allele] += 1


                pred_value = pred_row[col]

                # NaN = miscall
                if pd.isna(pred_value):

                    miscalled_calls += 1

                    for allele in gs_alleles:
                        miscalled_alleles[allele] += 1

                    continue

                pred_value = str(pred_value).strip()

                # Empty / invalid prediction = miscall
                if pred_value in {
                    "",
                    ".",
                    "nan",
                    "NaN",
                    "None",
                }:

                    miscalled_calls += 1

                    for allele in gs_alleles:
                        miscalled_alleles[allele] += 1

                    continue

                predicted_alleles = {
                    allele.strip()
                    for allele in pred_value.split("/")
                    if allele.strip()
                    and allele.strip() not in {
                        ".",
                        "nan",
                        "NaN",
                        "None",
                    }
                }

                # if at least one predicted allele matches at least one GS allele, it's a correct call
                is_correct = False

                for gs_allele in gs_alleles:
                    for predicted_allele in predicted_alleles:

                        gs_normalized, pred_normalized = (
                            normalize_allele_resolution(
                                gs_allele,
                                predicted_allele,
                            )
                        )

                        if gs_normalized == pred_normalized:
                            is_correct = True
                            break

                    if is_correct:
                        break

                if is_correct:
                    continue

                miscalled_calls += 1

                for allele in gs_alleles:
                    miscalled_alleles[allele] += 1

    return (
        miscalled_alleles,
        allele_opportunities,
        total_calls,
        miscalled_calls,
    )
```

```python
import math
from collections import Counter
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


dataset_numbers = sorted({
    int(path.stem.rsplit("_d", 1)[1])
    for path in results_folder.glob("*_d*.csv")
})


# ---------------------------------------------------------
# Create one figure containing all datasets
# ---------------------------------------------------------

n_datasets = len(dataset_numbers)
n_cols = 4
n_rows = math.ceil(n_datasets / n_cols)

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(16, 5 * n_rows),
    squeeze=False
)

axes = axes.flatten()


for plot_idx, dataset_number in enumerate(dataset_numbers):

    ax = axes[plot_idx]

    gs_path = datasets_folder / f"{dataset_number}_gs.csv"

    if not gs_path.exists():
        raise FileNotFoundError(
            f"No gold standard file found for dataset {dataset_number}: "
            f"{dataset_number}: {gs_path}"
        )

    gs = pd.read_csv(
        gs_path,
        dtype=str
    )


    miscalled_counts = Counter()
    opportunity_counts = Counter()

    total_calls = 0
    total_miscalls = 0


    result_paths = sorted(
        results_folder.glob(
            f"*_d{dataset_number}.csv"
        )
    )

    if not result_paths:
        print(
            f"No prediction files found for dataset {dataset_number}."
        )
        ax.set_visible(False)
        continue


    # ---------------------------------------------------------
    # Aggregate all tools
    # ---------------------------------------------------------

    for result_path in result_paths:

        predictions = pd.read_csv(
            result_path,
            dtype=str
        )

        tool_name = result_path.stem.rsplit(
            "_d",
            1
        )[0]


        (
            miscalled,
            opportunities,
            n_calls,
            n_miscalls,
        ) = get_miscalled_and_all_alleles(
            predictions,
            gs,
            tool_name,
            tool_predicted_alleles,
        )


        miscalled_counts.update(
            miscalled
        )

        opportunity_counts.update(
            opportunities
        )

        total_calls += n_calls
        total_miscalls += n_miscalls


    # ---------------------------------------------------------
    # Remove invalid alleles
    # ---------------------------------------------------------

    invalid = {
        "",
        ".",
        "nan",
        "NaN",
        "None",
        "ERR",
    }


    miscalled_counts = Counter({
        allele: count
        for allele, count in miscalled_counts.items()
        if (
            pd.notna(allele)
            and str(allele).strip()
            not in invalid
        )
    })


    opportunity_counts = Counter({
        allele: count
        for allele, count in opportunity_counts.items()
        if (
            pd.notna(allele)
            and str(allele).strip()
            not in invalid
        )
    })


    # ---------------------------------------------------------
    # Top 20 most frequently miscalled alleles
    # ---------------------------------------------------------

    top_20 = miscalled_counts.most_common(20)


    if not top_20:
        print(
            f"Dataset {dataset_number}: "
            "no incorrectly called alleles."
        )

        ax.set_visible(False)
        continue


    alleles = []
    miscalls = []
    opportunities = []
    percentages = []


    for allele, miscall_count in top_20:

        opportunity_count = opportunity_counts.get(
            allele,
            0
        )

        if opportunity_count == 0:
            continue


        rate = (
            100
            * miscall_count
            / opportunity_count
        )


        alleles.append(allele)
        miscalls.append(miscall_count)
        opportunities.append(opportunity_count)
        percentages.append(rate)


    # ---------------------------------------------------------
    # Sort by percentage
    # Largest percentage at the TOP
    # ---------------------------------------------------------

    sorted_data = sorted(
        zip(
            alleles,
            miscalls,
            opportunities,
            percentages
        ),
        key=lambda x: x[3],
        reverse=True
    )


    # barh() puts the first element at the bottom,
    # so reverse the order
    alleles = [x[0] for x in sorted_data][::-1]
    miscalls = [x[1] for x in sorted_data][::-1]
    opportunities = [x[2] for x in sorted_data][::-1]
    percentages = [x[3] for x in sorted_data][::-1]


    # ---------------------------------------------------------
    # Colors by locus
    # ---------------------------------------------------------

    colors = [
        allele_colors.get(
            get_locus(allele),
            "gray"
        )
        for allele in alleles
    ]


    # ---------------------------------------------------------
    # Plot
    # ---------------------------------------------------------

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


    bars = ax.barh(
        alleles,
        percentages,
        color=colors
    )


    # Labels: miscalls / opportunities
    labels = [
        f"{mis}/{opp}"
        for mis, opp in zip(
            miscalls,
            opportunities
        )
    ]


    ax.bar_label(
        bars,
        labels=labels,
        padding=3,
        fontsize=8
    )


    ax.set_xlim(0, 100)
    ax.set_xticks(range(0, 101, 20))


    ax.set_xlabel(
        "% of calls mispredicted"
    )


    ax.set_title(
        f"Dataset {dataset_number}",
        fontsize=12,
        fontweight="bold"
    )


    # ---------------------------------------------------------
    # Legend for loci present in this dataset
    # ---------------------------------------------------------

    present_loci = []

    for allele in alleles:

        locus = get_locus(allele)

        if locus not in present_loci:
            present_loci.append(locus)


    legend_handles = [
        Patch(
            facecolor=allele_colors[locus],
            label=locus
        )
        for locus in allele_colors
        if locus in present_loci
    ]


    ax.legend(
        handles=legend_handles,
        title="Locus",
        loc="lower right",
        frameon=True
    )

fig.suptitle(
    "Top 20 Most-Frequently Miscalled Alleles",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout(
    rect=[0, 0, 1, 0.96]
)

plt.show()
```
